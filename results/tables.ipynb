# Tables

```python
import pandas as pd
```

```python
def get_pairs(df: pd.DataFrame) -> set:
    return {(r.upstream, r.downstream) for r in df.itertuples()}
```

```python
# Load data
as_hegemony_v4 = pd.read_csv('as-pairs/as-hegemony/single-upstream-pairs-as-hegemony-v4.csv')
as_hegemony_v6 = pd.read_csv('as-pairs/as-hegemony/single-upstream-pairs-as-hegemony-v6.csv')
as_relationship_v4 = pd.read_csv(
    'as-pairs/as-relationship/single-upstream-pairs-as-relationship-v4.csv'
)
as_relationship_v6 = pd.read_csv(
    'as-pairs/as-relationship/single-upstream-pairs-as-relationship-v6.csv'
)
measured_v4 = pd.concat(
    [
        pd.read_csv('as-pairs/measured/backup-upstream-pairs-measured-v4.csv'),
        pd.read_csv('as-pairs/measured/default-route-pairs-measured-v4.csv'),
        pd.read_csv('as-pairs/measured/single-upstream-pairs-measured-v4.csv'),
    ]
)
measured_v6 = pd.concat(
    [
        pd.read_csv('as-pairs/measured/backup-upstream-pairs-measured-v6.csv'),
        pd.read_csv('as-pairs/measured/default-route-pairs-measured-v6.csv'),
        pd.read_csv('as-pairs/measured/single-upstream-pairs-measured-v6.csv'),
    ]
)
direct_pairs_v4 = pd.read_csv('as-pairs/as-hegemony/direct-pairs-v4.csv')
indirect_pairs_v4 = pd.read_csv('as-pairs/as-hegemony/indirect-pairs-v4.csv')
direct_pairs_v6 = pd.read_csv('as-pairs/as-hegemony/direct-pairs-v6.csv')
indirect_pairs_v6 = pd.read_csv('as-pairs/as-hegemony/indirect-pairs-v6.csv')
indirect_pairs_in_customer_cone_v4 = pd.read_csv(
    'as-pairs/as-hegemony/customer-cone/indirect-pairs-in-customer-cone-v4.csv'
)
indirect_pairs_not_in_customer_cone_v4 = pd.read_csv(
    'as-pairs/as-hegemony/customer-cone/indirect-pairs-not-in-customer-cone-v4.csv'
)
indirect_pairs_no_data_v4 = pd.read_csv(
    'as-pairs/as-hegemony/customer-cone/indirect-pairs-no-data-v4.csv'
)
downstream_dependency_ip_version = pd.read_csv('as-hegemony/downstream-dependency-ip-version.csv')
dual_stack_downstreams_upstream_relationship = pd.read_csv(
    'as-hegemony/dual-stack-downstreams-upstream-relationships.csv'
)
sibling_classification_as_hegemony_v4 = pd.read_csv(
    'as-hegemony/sibling-classification-as-hegemony-v4.csv'
)
sibling_classification_measured_v4 = pd.read_csv('measured/sibling-classification-measured-v4.csv')
single_upstream_pairs_measured_v4 = pd.read_csv('as-pairs/measured/single-upstream-pairs-measured-v4.csv')
single_upstream_pairs_isp_v4 = get_pairs(
    pd.read_csv('as-pairs/measured/single-upstream-pairs-measured-v4.csv')
).intersection(get_pairs(pd.read_csv('as-pairs/measured/isp-pairs-v4.csv')))
sibling_classification_isp_v4 = sibling_classification_measured_v4[
    sibling_classification_measured_v4.apply(
        lambda r: (r['upstream'], r['downstream']) in single_upstream_pairs_isp_v4, axis='columns'
    )
]
downstreams_at_ixps_as_hegemony_v4 = pd.read_csv('as-hegemony/downstreams-at-ixps-as-hegemony-v4.csv')
hg_reachable_via_ixp_as_hegemony_v4 = pd.read_csv('as-hegemony/hg-reachable-via-ixp-as-hegemony-v4.csv')
downstreams_at_ixps_measured_v4 = pd.read_csv('measured/downstreams-at-ixps-measured-v4.csv')
hg_reachable_via_ixp_measured_v4 = pd.read_csv('measured/hg-reachable-via-ixp-measured-v4.csv')
```

## Table 1

```python
symmetric_difference_v4 = pd.concat([as_hegemony_v4, as_relationship_v4], keys=['as_hegemony', 'as_relationship']).drop_duplicates(keep=False)
symmetric_difference_v6 = pd.concat([as_hegemony_v6, as_relationship_v6], keys=['as_hegemony', 'as_relationship']).drop_duplicates(keep=False)
table_1 = pd.DataFrame(
    [
        [
            'AS Hegemony',
            as_hegemony_v4['upstream'].nunique(),
            as_hegemony_v6['upstream'].nunique(),
            as_hegemony_v4['downstream'].nunique(),
            as_hegemony_v6['downstream'].nunique(),
            len(as_hegemony_v4),
            len(as_hegemony_v6),
        ],
        [
            'AS relationship',
            as_relationship_v4['upstream'].nunique(),
            as_relationship_v6['upstream'].nunique(),
            as_relationship_v4['downstream'].nunique(),
            as_relationship_v6['downstream'].nunique(),
            len(as_relationship_v4),
            len(as_relationship_v6),
        ],
        [
            'Overlap',
            len(set(as_hegemony_v4['upstream']).intersection(as_relationship_v4['upstream'])),
            len(set(as_hegemony_v6['upstream']).intersection(as_relationship_v6['upstream'])),
            len(set(as_hegemony_v4['downstream']).intersection(as_relationship_v4['downstream'])),
            len(set(as_hegemony_v6['downstream']).intersection(as_relationship_v6['downstream'])),
            len(pd.merge(as_hegemony_v4, as_relationship_v4, how='inner')),
            len(pd.merge(as_hegemony_v6, as_relationship_v6, how='inner')),
        ],
        [
            'AS Hegemony only',
            len(set(as_hegemony_v4['upstream']).difference(as_relationship_v4['upstream'])),
            len(set(as_hegemony_v6['upstream']).difference(as_relationship_v6['upstream'])),
            len(set(as_hegemony_v4['downstream']).difference(as_relationship_v4['downstream'])),
            len(set(as_hegemony_v6['downstream']).difference(as_relationship_v6['downstream'])),
            len(symmetric_difference_v4.loc['as_hegemony']),
            len(symmetric_difference_v6.loc['as_hegemony']),
        ],
        [
            'AS relationship only',
            len(set(as_relationship_v4['upstream']).difference(as_hegemony_v4['upstream'])),
            len(set(as_relationship_v6['upstream']).difference(as_hegemony_v6['upstream'])),
            len(set(as_relationship_v4['downstream']).difference(as_hegemony_v4['downstream'])),
            len(set(as_relationship_v6['downstream']).difference(as_hegemony_v6['downstream'])),
            len(symmetric_difference_v4.loc['as_relationship']),
            len(symmetric_difference_v6.loc['as_relationship']),
        ],
        [
            'Measured',
            measured_v4['upstream'].nunique(),
            measured_v6['upstream'].nunique(),
            measured_v4['downstream'].nunique(),
            measured_v6['downstream'].nunique(),
            len(measured_v4),
            len(measured_v6),
        ],
    ],
    columns=[
        '',
        'Upstreams (IPv4)',
        'Upstreams (IPv6)',
        'Downstreams (IPv4)',
        'Downstreams (IPv6)',
        'Pairs (IPv4)',
        'Pairs (IPv6)',
    ],
)
table_1
```

## Table 3

```python
single_stack_v4 = downstream_dependency_ip_version[downstream_dependency_ip_version['stack'] == 'single-v4']
single_stack_v6 = downstream_dependency_ip_version[downstream_dependency_ip_version['stack'] == 'single-v6']
dual_stack = downstream_dependency_ip_version[downstream_dependency_ip_version['stack'] == 'dual']
table_3 = pd.DataFrame(
    [
        [
            'Single stack IPv4',
            len(single_stack_v4),
            pd.NA,
            pd.NA,
            len(single_stack_v4),
        ],
        [
            'Single stack IPv6',
            pd.NA,
            len(single_stack_v6),
            pd.NA,
            len(single_stack_v6),
        ],
        [
            'Dual stack',
            len(dual_stack[dual_stack['dependency'] == 'v4-only']),
            len(dual_stack[dual_stack['dependency'] == 'v6-only']),
            len(dual_stack[dual_stack['dependency'] == 'both']),
            len(dual_stack)
        ],
        [
            'Total',
            len(downstream_dependency_ip_version[downstream_dependency_ip_version['dependency'] == 'v4-only']),
            len(downstream_dependency_ip_version[downstream_dependency_ip_version['dependency'] == 'v6-only']),
            len(downstream_dependency_ip_version[downstream_dependency_ip_version['dependency'] == 'both']),
            len(downstream_dependency_ip_version)
        ],
    ],
    columns=['', 'IPv4-only', 'IPv6-only', 'Both', 'Total']
)
table_3
```

```python
num_su_downstreams_as_hegemony = as_hegemony_v4['downstream'].nunique()
num_su_downstreams_measured = single_upstream_pairs_measured_v4['downstream'].nunique()
su_downstreams_isp = {e[1] for e in single_upstream_pairs_isp_v4}
num_su_downstreams_isp = len(su_downstreams_isp)
downstreams_at_ixps_isp_v4 = downstreams_at_ixps_measured_v4[
    downstreams_at_ixps_measured_v4['downstream'].isin(su_downstreams_isp)
]
hg_reachable_via_ixp_isp_v4 = hg_reachable_via_ixp_measured_v4[
    hg_reachable_via_ixp_measured_v4['downstream'].isin(su_downstreams_isp)
]
table_6 = pd.DataFrame(
    [
        [
            'BGP view',
            num_su_downstreams_as_hegemony,
            downstreams_at_ixps_as_hegemony_v4['downstream'].nunique(),
            downstreams_at_ixps_as_hegemony_v4['downstream'].nunique()
            / num_su_downstreams_as_hegemony
            * 100,
            hg_reachable_via_ixp_as_hegemony_v4['downstream'].nunique(),
            hg_reachable_via_ixp_as_hegemony_v4['downstream'].nunique()
            / num_su_downstreams_as_hegemony
            * 100,
        ],
        [
            'Measured',
            num_su_downstreams_measured,
            downstreams_at_ixps_measured_v4['downstream'].nunique(),
            downstreams_at_ixps_measured_v4['downstream'].nunique()
            / num_su_downstreams_measured
            * 100,
            hg_reachable_via_ixp_measured_v4['downstream'].nunique(),
            hg_reachable_via_ixp_measured_v4['downstream'].nunique()
            / num_su_downstreams_measured
            * 100,
        ],
        [
            'ISPs',
            num_su_downstreams_isp,
            downstreams_at_ixps_isp_v4['downstream'].nunique(),
            downstreams_at_ixps_isp_v4['downstream'].nunique() / num_su_downstreams_isp * 100,
            hg_reachable_via_ixp_isp_v4['downstream'].nunique(),
            hg_reachable_via_ixp_isp_v4['downstream'].nunique() / num_su_downstreams_isp * 100,
        ],
    ],
    columns=['Source', 'Total', '...at IXPs', '(%)', '...HG peer', '(%)'],
)
table_6
```
